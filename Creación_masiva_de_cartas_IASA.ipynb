```python
import pandas as pd
from docx import Document

# ===== Leer Excel =====
df = pd.read_excel("asignaciones (1).xlsx")

# Renombrar columnas para comodidad
df = df.rename(columns={
    "SALON ": "SALON",
    "APLICADOR (A)": "APLICADOR",
    "Unnamed: 6": "GENERO"
})

df
```

```python
from docx.enum.text import WD_ALIGN_PARAGRAPH

# ===== Agrupar por aplicador =====
for aplicador, datos in df.groupby("APLICADOR"):

    genero = datos["GENERO"].iloc[0]

    # ---- Saludo según género ----
    if genero == "H":
        saludo = f"Estimado Mtro. {aplicador},"
    else:
        saludo = f"Estimada Mtra. {aplicador},"

    doc = Document()

    # ===== ENCABEZADO =====

    encabezado = [
        "ESCUELA NACIONAL PREPARATORIA",
        'No. 4 "VIDAL CASTAÑEDA Y NÁJERA"',
        "SECRETARÍA GENERAL"
        ]

    for linea in encabezado:
        p = doc.add_paragraph()
        run = p.add_run(linea)
        run.bold = True
        p.alignment = WD_ALIGN_PARAGRAPH.RIGHT


    # ===== DEPENDENCIA =====
    p2 = doc.add_paragraph()
    run2 = p2.add_run(
        'ESCUELA NACIONAL PREPARATORIA\n'
        'No. 4 "VIDAL CASTAÑEDA Y NÁJERA"\n'
        'P r e s e n t e'
    )
    run2.bold = True


    # ===== SALUDO =====
    p3 = doc.add_paragraph()
    run3 = p3.add_run(saludo)
    run3.bold = True

    # ===== TEXTO =====
    p4 = doc.add_paragraph()

    run4 = p4.add_run( "Por medio de la presente, se le informa que ha sido designado(a) "
            "como aplicador(a) de la evaluación del Instrumento de Apoyo a la "
            "Superación Académica (IASA-2026).")
    p4.alignment = WD_ALIGN_PARAGRAPH.JUSTIFY

    p5= doc.add_paragraph()

    run5 = p5.add_run( "A continuación, se detalla la programación de los grupos que deberá "
            "acompañar al centro de cómputo correspondiente:")
    p5.alignment = WD_ALIGN_PARAGRAPH.JUSTIFY

    # ===== TABLA DE PROGRAMACIÓN =====
    tabla = doc.add_table(rows=1, cols=5)
    encabezado = tabla.rows[0].cells

    encabezado[0].text = "Día"
    encabezado[1].text = "Horario"
    encabezado[2].text = "Grupo"
    encabezado[3].text = "Salón"
    encabezado[4].text = "Centro"

    for _, fila in datos.iterrows():
        celdas = tabla.add_row().cells
        celdas[0].text = str(fila["DIA"])
        celdas[1].text = str(fila["HORA"])
        celdas[2].text = str(fila["GRUPO"])
        celdas[3].text = str(fila["SALON"])
        celdas[4].text = str(fila["CENTRO"])

    # ===== INSTRUCCIONES =====
    p6 = doc.add_paragraph()

    run6 = p6.add_run( "\nSe solicita presentarse puntualmente en el salón indicado para "
        "recoger al grupo y trasladarlo al centro de cómputo asignado, así "
        "como permanecer con los estudiantes hasta la conclusión de la evaluación.")
    p6.alignment = WD_ALIGN_PARAGRAPH.JUSTIFY

    p7 = doc.add_paragraph ()
    run7 = p7.add_run("Agradecemos de antemano su valiosa colaboración.")
    p7.alignment = WD_ALIGN_PARAGRAPH.JUSTIFY

    #====== CIerre =======

    cierre = [
    "A t e n t a m e n t e",
    "“POR MI RAZA HABLARÁ EL ESPÍRITU”",
    "Tacubaya, Ciudad de México, ___ de ______ de 2026",
    "SECRETARÍA GENERAL",
    "M. en E. Martha Marín Pérez"]


    # ===== GUARDAR =====
    for linea in cierre:
      p = doc.add_paragraph()
      run = p.add_run(linea)
      run.bold = True
      p.alignment = WD_ALIGN_PARAGRAPH.LEFT

    nombre_archivo = f"Carta_{aplicador}.docx"
    doc.save(nombre_archivo)
```
